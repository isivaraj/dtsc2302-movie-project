**Research question:** Do movies with higher budgets tend to receive higher audience ratings and earn more revenue?

This notebook starts collecting data from The Movie Database (TMDB). The goal is to create a table of movie titles, budgets, revenues, audience ratings, release dates, and genres.

**Note:** You will need a free TMDB API key. Do not put your API key in a public GitHub repository.

```python
import requests
import csv
from getpass import getpass
```

```python
api_key = getpass("Enter your TMDB API key (hidden): ")
base_url = "https://api.themoviedb.org/3"
```

```python
response = requests.get(
    f"{base_url}/movie/popular",
    params={"api_key": api_key, "language": "en-US", "page": 1},
    timeout=20
)
response.raise_for_status()
popular_movies = response.json()["results"]
print("Movies found:", len(popular_movies))
```

```python
movies = []

for movie in popular_movies:
    movie_id = movie["id"]
    response = requests.get(
        f"{base_url}/movie/{movie_id}",
        params={"api_key": api_key},
        timeout=20
    )
    response.raise_for_status()
    details = response.json()

    movies.append({
        "title": details.get("title"),
        "budget": details.get("budget"),
        "revenue": details.get("revenue"),
        "audience_rating": details.get("vote_average"),
        "vote_count": details.get("vote_count"),
        "release_date": details.get("release_date"),
        "genres": ", ".join(g["name"] for g in details.get("genres", []))
    })

print("Movie records collected:", len(movies))
print(movies[:2])
```

```python
with open("movies.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=movies[0].keys())
    writer.writeheader()
    writer.writerows(movies)

print("Saved movies.csv")
```
